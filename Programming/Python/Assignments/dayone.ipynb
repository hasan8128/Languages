```python
def median(data):
    data.sort()
    n = len(data)

    if n % 2 == 1:
        return data[n //2]
    else:
        middle1 = data[n//2 -1]
        middle2 = data[n//2]
        return (middle1 + middle2)/2

data = [1,2, 3, 4, 5]

print(median(data))

print(median([12, 7, 25, 9, 18, 30]))
```

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print(np.mean(data))
print(np.median(data))
print(np.var(data))
print(np.std(data))
```

```python
import math

def variance(data):
    mean = sum(data)/len(data)

    squared_deviations = []

    for x in data:
        deviation = x - mean
        squared = deviation**2
        squared_deviations.append(squared)

    return sum(squared_deviations)/len(data)
    


def standard_deviation(data):
    return math.sqrt(variance(data))

data = [10, 20, 30]

print(variance(data))
print(standard_deviation(data))

```
