```python
class Person:
    def __init__(self, name, age):
        self.name = name  # 'self.name' is attached to the object
        self.age = age    # 'self.age' is also attached to the object

    def show_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

p = Person("Alice", 25)
p.show_info()
```
